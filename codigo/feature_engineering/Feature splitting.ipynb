#### se crean las variables año_publicacion y año_apertura

```python
#AÑO publicación en el portal
data_procesos['año_publicacion'] = data_procesos['cro_fecha_publicacion'].astype(str).str[6:10] 
data_procesos['año_publicacion']  = data_procesos['año_publicacion'].str.replace('tos','sin datos')
#AÑO apertura en el portal
data_procesos['año_apertura'] = data_procesos['fecha_apertura'].astype(str).str[6:10] 
data_procesos['año_apertura']  = data_procesos['año_apertura'].str.replace('tos','sin datos')
```

#### crea variables periodo a partir de fecha

```python
#crea variables periodo a partir de fecha
data_procesos['periodo_apertura'] = data_procesos['fecha_apertura'].astype(str).str[6:10] + data_procesos['fecha_apertura'].astype(str).str[3:5] 
data_procesos['periodo_apertura']  = data_procesos['periodo_apertura'].str.replace('tos d','sin datos')

data_procesos['periodo_publicacion'] = data_procesos['cro_fecha_publicacion'].astype(str).str[6:10] + data_procesos['cro_fecha_publicacion'].astype(str).str[3:5] 
data_procesos['periodo_publicacion']  = data_procesos['periodo_publicacion'].str.replace('tos d','sin datos')

data_procesos['periodo_inicio_consultas'] = data_procesos['cro_fecha_inicio_consultas'].astype(str).str[6:10] + data_procesos['cro_fecha_inicio_consultas'].astype(str).str[3:5] 
data_procesos['periodo_inicio_consultas']  = data_procesos['periodo_inicio_consultas'].str.replace('tos d','sin datos')

data_procesos['periodo_final_consultas'] = data_procesos['cro_fecha_final_consultas'].astype(str).str[6:10] + data_procesos['cro_fecha_final_consultas'].astype(str).str[3:5] 
data_procesos['periodo_final_consultas']  = data_procesos['periodo_final_consultas'].str.replace('tos d','sin datos')

data_procesos['periodo_inicio_recepcion_documentos'] = data_procesos['cro_fecha_inicio_recepcion_documentos'].astype(str).str[6:10] + data_procesos['cro_fecha_inicio_recepcion_documentos'].astype(str).str[3:5] 
data_procesos['periodo_inicio_recepcion_documentos']  = data_procesos['periodo_inicio_recepcion_documentos'].str.replace('tos d','sin datos')

data_procesos['periodo_fin_recepcion_documentos'] = data_procesos['cro_fecha_fin_recepcion_documentos'].astype(str).str[6:10] + data_procesos['cro_fecha_fin_recepcion_documentos'].astype(str).str[3:5] 
data_procesos['periodo_fin_recepcion_documentos']  = data_procesos['periodo_fin_recepcion_documentos'].str.replace('tos d','sin datos')

data_procesos['periodo_acto_apertura'] = data_procesos['cro_fecha_acto_apertura'].astype(str).str[6:10] + data_procesos['cro_fecha_acto_apertura'].astype(str).str[3:5] 
data_procesos['periodo_acto_apertura']  = data_procesos['periodo_acto_apertura'].str.replace('tos d','sin datos')
```
