## Procesos

```python
data_procesos=pd.read_csv(ruta_dataset_feature_creation+"procesos2016_2022_nuevas_var.csv", sep='|', encoding='utf-8-sig')
```

```python
data_procesos
```

```python
#se elimina columna
data_procesos=data_procesos.drop(['Unnamed: 0.1'], axis=1)
```

#### discretización de proveedores participantes

```python
#se discretiza la variable proveedores participantes y para ello se construyen veintiles 
data_procesos['d_proveedores_participantes'] = pd.qcut(data_procesos['proveedores_participantes'], q=20,duplicates='drop')
data_procesos['d_proveedores_participantes'].describe()
```

```python
datos=data_procesos.groupby(['d_proveedores_participantes']).size().reset_index(name='Nro procesos')
datos['total']=data_procesos['d_proveedores_participantes'].count()
datos
```

#### discretización de ofertas confirmadas

```python
data_procesos['d_ofertas_confirmadas'] = pd.qcut(data_procesos['ofertas_confirmadas'], q=20,duplicates='drop')
#pd.qcut(s,len(s.dropna()),duplicates='drop')
data_procesos['d_ofertas_confirmadas'].describe()
```

```python
datos=data_procesos.groupby(['d_ofertas_confirmadas']).size().reset_index(name='Nro procesos')
datos['total']=data_procesos['d_ofertas_confirmadas'].count()
datos
```

#### discretización de días de recepción de documentos en soporte físico

```python
data_procesos['dias_recepción_doc_soporte_fisico']=data_procesos['dias_recepción_doc_soporte_fisico'].replace('sin datos', np.nan)
data_procesos['dias_recepción_doc_soporte_fisico']=pd.to_numeric(data_procesos['dias_recepción_doc_soporte_fisico'])
data_procesos['d_dias_recepcion_doc_soporte_fisico'] = pd.qcut(data_procesos['dias_recepción_doc_soporte_fisico'], q=20,duplicates='drop')
#pd.qcut(s,len(s.dropna()),duplicates='drop')
data_procesos['d_dias_recepcion_doc_soporte_fisico'].describe()
```

#### discretización de cantidad total de bienes

```python
data_procesos['d_cant_total_bienes'] = pd.qcut(data_procesos['cant_total_bienes'], q=20,duplicates='drop')
#pd.qcut(s,len(s.dropna()),duplicates='drop')
data_procesos['d_cant_total_bienes'].describe()
```

```python
datos=data_procesos.groupby(['d_cant_total_bienes']).size().reset_index(name='Nro procesos')
datos['total']=data_procesos['d_cant_total_bienes'].count()
datos
```

#### discretización de cantidad total de servicios

```python
data_procesos['d_cant_total_servicios'] = pd.qcut(data_procesos['cant_total_servicios'], q=20,duplicates='drop')
data_procesos['d_cant_total_servicios'].describe()
```

```python
datos=data_procesos.groupby(['d_cant_total_servicios']).size().reset_index(name='Nro procesos')
datos['total']=data_procesos['d_cant_total_servicios'].count()
datos
```

#### discretización de monto total de procesos

```python
data_procesos['dP_monto_total_procesos'] = pd.qcut(data_procesos['P_monto_total_procesos'], q=20,duplicates='drop')
data_procesos['dP_monto_total_procesos'].describe()
```

```python
datos=data_procesos.groupby(['dP_monto_total_procesos']).size().reset_index(name='Nro procesos')
datos['total']=data_procesos['dP_monto_total_procesos'].count()
datos
```

#### discretización de cantidad de proveedores

```python
data_procesos['dP_cant_proveedores'] = pd.qcut(data_procesos['P_cant_proveedores'], q=20,duplicates='drop')
data_procesos['dP_cant_proveedores'].describe()
```

```python
datos=data_procesos.groupby(['dP_cant_proveedores']).size().reset_index(name='Nro procesos')
datos['total']=data_procesos['dP_cant_proveedores'].count()
datos
```

#### discretización de duración de contrato en días

```python
data_procesos['d_duracion_contrato_dias']=data_procesos['duracion_contrato_dias'].replace('sin datos', np.nan)
data_procesos['d_duracion_contrato_dias']=pd.to_numeric(data_procesos['duracion_contrato_dias'])
data_procesos['d_duracion_contrato_dias'] = pd.qcut(data_procesos['duracion_contrato_dias'], q=20,duplicates='drop')
#pd.qcut(s,len(s.dropna()),duplicates='drop')
data_procesos['d_duracion_contrato_dias'].describe()
```

```python
datos=data_procesos.groupby(['d_duracion_contrato_dias']).size().reset_index(name='Nro procesos')
datos['total']=data_procesos['d_duracion_contrato_dias'].count()
datos
```
