### Armar heatmap

```python
# Calculamos el porcentaje de valores nulos por columna
porcentaje_nulos = data_procesos.isnull().mean()

# Seleccionamos las columnas con al menos 50% de nulos
columnas_con_muchos_nulos = porcentaje_nulos[porcentaje_nulos >= 0.5].index

# Creamos un nuevo DataFrame solo con esas columnas
df_nulos_50 = data_procesos[columnas_con_muchos_nulos]
```

```python
# Heatmap de correlación de missing values
plt.figure(figsize=(2, 1))
msno.heatmap(df_nulos_50, fontsize=14)
plt.title("Correlación entre valores faltantes", fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.tight_layout()
plt.show()
```
